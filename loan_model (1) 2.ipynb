```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('Smartphone_Usage_Productivity_Dataset_50000.csv')
```

```python
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
```

```python
encoder_Gender = LabelEncoder()
encoder_Occupation = LabelEncoder()
encoder_Device_Type = LabelEncoder()



data['Genderr'] = encoder_Gender.fit_transform(data['Gender'])

data['Occupation'] = encoder_Occupation.fit_transform(data['Occupation'])

data['Device_Type'] = encoder_Device_Type.fit_transform(data['Device_Type'])
```

```python
print(encoder_Gender.classes_,encoder_Gender.transform(encoder_Gender.classes_))
print(encoder_Occupation.classes_,encoder_Occupation.transform(encoder_Occupation.classes_))
print(encoder_Device_Type.classes_,encoder_Device_Type.transform(encoder_Device_Type.classes_))
```

```python
data.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
X = data.drop(columns=['Stress_Level','User_ID'])
Y = data['Stress_Level']
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2)
```

```python
model = LogisticRegression(max_iter=10000)
```

```python
model.fit(X_train.drop(columns=['Gender']),Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
acc = accuracy_score(Y_test,Y_pred)
print(acc)
```
